# Telecomunicaciones: identificar operadores ineficaces

CallMeMaybe es un servicio de telefonía virtual, este tipo de servicio utiliza la tecnología de Internet para transmitir llamadas telefónicas. A diferencia de los sistemas telefónicos tradicionales, que requieren infraestructura física, la telefonía virtual opera en la nube, permitiendo a las empresas gestionar sus comunicaciones de manera más eficiente y económica.
Los beneficios que ofrece es flexibilidad, movilidad y la posibilidad de realizar llamadas desde cualquier dispositivo con conexión a Internet. Se utiliza en empresas para mejorar la comunicación interna y externa, facilitando la atención al cliente y la gestión de llamadas.

El objetivo es identificar a operadores ineficientes dentro de este servicio. De esta manera, se debe tomar acción sobre ellos para incrementar su desempeño y rendimiento. Por lo que se estudiará el comportamiento de cada operador junto con las actividades que realiza; llamadas entrantes o salientes, tiempo de espera, llamadas perdidas, duración de la llamada, entre otras.. Esta información ayudará a analizar su trabajo y evaluarlo.

Una vez que se ha identificado a los operadores con el menor desempeño, se puede optar por capacitarlos, su reasignación a otras áreas o mejorar el servicio al cliente. Lo importante es corregir este problema para evitar que siga afectando el servicio que ofrece la compañía CallMeMaybe. Por efecto, la calidad del servicio se verá en aumento significando un gran beneficio que atraerá más clientes buscando el servicio eficaz que ofrece, posicionandose como uno de los mejores servicios de telefonía virtual. 

## **Descripción de los datos**

Los datasets contienen información sobre el uso del servicio de telefonía virtual CallMeMaybe. Sus clientes son organizaciones que necesitan distribuir gran cantidad de llamadas entrantes entre varios operadores, o realizar llamadas salientes a través de sus operadores. Los operadores también pueden realizar llamadas internas para comunicarse entre ellos. Estas llamadas se realizan a través de la red de CallMeMaybe. 

**Dataset 1: telecom_dataset_us.csv**

Este es el dataset operativo, de el se obtendran los datos que ayudarán a medir el desempeño, contar llamadas, calcular tiempos y evaluar comportamiento por operador.
Contiene 53902 filas con 9 columnas, contiene valores perdidos en algunas columnas y otras se tendrán que convertir a otro tipo de datos.

**Dataset 2: telecom_clients_us.csv**

Sirve para entender el tipo de cliente, segmentar por tarifa y analizar si ciertos clientes tienen más operadores ineficaces. Contiene 732 filas con 3 columnas, la columnas date_start se convertirá al tipo fecha. Si bien, contiene información valiosa, no se ocupará con la misma frecuencia que el primer dataset. 

#### **Resumen estadístico**

Como parte del análisis exploratorio, se realizará un resumen estadístico de las principales variables relacionadas con el desempeño de los operadores.

En primer lugar, se analizará la actividad de llamadas por operador, considerando estadísticas de centralidad y dispersión del número de llamadas, con el objetivo de identificar operadores con niveles de actividad significativamente bajos.

Posteriormente, se estudiará el comportamiento de las llamadas perdidas, contabilizando su frecuencia y proporción por operador, así como su distribución, para detectar posibles patrones de ineficiencia.

Asimismo, se analizarán la duración de las llamadas y el tiempo de espera, utilizando medidas como la media, mediana y desviación estándar, con el fin de evaluar la variabilidad en la atención de llamadas entre operadores.

De manera complementaria, se compararán las llamadas internas y externas, así como las llamadas entrantes y salientes, para identificar diferencias relevantes en el comportamiento de los operadores.

Finalmente, se analizará la presencia de valores atípicos en variables clave, como el número de llamadas perdidas y los tiempos de espera, con el objetivo de detectar comportamientos extremos que puedan indicar posibles casos de ineficiencia.

También se explorará la dimensión temporal de los datos para identificar si existen patrones asociados a determinadas fechas o periodos.


#### **Las variables con mayor interés son las siguientes:**

* _is_missed_call_: permite identificar la cantidad de llamadas perdidas, uno de los principales criterios para definir la ineficacia de un operador.

* _total_call_duration_: incluye el tiempo de espera, lo cual es clave para evaluar si los operadores generan demoras prolongadas en las llamadas entrantes.

* _call_duration_: representa la duración efectiva de las llamadas atendidas y ayuda a analizar el manejo del tiempo por operador.

* _calls_count_: indica el volumen de llamadas gestionadas y permite detectar operadores con baja actividad.

* _direction_: diferencia entre llamadas entrantes y salientes, lo que es fundamental para evaluar el desempeño esperado de cada operador.

* _operator_id_: permite identificar y analizar el comportamiento individual de cada operador.

* _internal_: distingue llamadas internas y externas, lo cual ayuda a enfocar el análisis en la atención al cliente.

Estas variables son esenciales para comprender el comportamiento y la eficiencia de los operadores dentro del servicio.

#### **Variables útiles para agregación:**

La agregación de datos a partir de las siguientes variables permitirá analizar el desempeño de los operadores a lo largo del tiempo y por cliente:

* _operator_id_: para evaluar el desempeño individual de cada operador.

* _user_id_: para identificar a las organizaciones afectadas.

* _date_: para analizar patrones temporales y evolución del desempeño.


## **Hipótesis a validar**

**Hipótesis 1**

H₀: No existe una diferencia significativa en la cantidad de llamadas perdidas entre operadores eficientes e ineficientes.

H₁: Los operadores ineficientes presentan una cantidad significativamente mayor de llamadas perdidas, tanto internas como externas.

**Hipótesis 2**

H₀: No existen diferencias significativas en los tiempos de espera de las llamadas entrantes entre operadores eficientes e ineficientes.

H₁: Los operadores ineficientes presentan tiempos de espera significativamente mayores en las llamadas entrantes.

**Hipótesis 3**

H₀: No existe una diferencia significativa en la cantidad de llamadas salientes realizadas por operadores eficientes e ineficientes.

H₁: Los operadores cuya función incluye realizar llamadas salientes y que presentan una cantidad significativamente reducida de estas llamadas pueden considerarse ineficientes.

## **Indicadores clave (KPIs)**

Con el objetivo de identificar operadores ineficientes dentro del servicio de telefonía CallMeMaybe, se definieron los siguientes indicadores KPIs. Estos indicadores permiten medir de manera objetiva el desempeño de cada operador a partir de las variables disponibles en el conjunto de datos y sirven como base para la validación de las hipótesis planteadas.

1. **Tasa de llamadas pérdidas por operador.**

   Mide la proporción de llamadas entrantes que no fueron atendidas por cada operador. Un valor   elevado puede reflejar una sobrecarga de trabajo, falta de disponibilidad o baja eficiencia operativa.

   Tasa de llamadas perdidas = Llamadas entrantes perdidas / Total de llamadas entrantes

2. **Tiempo promedio de espera.**

   Este KPI permite evaluar cuánto tiempo, en promedio, los clientes deben esperar antes de que sus llamadas entrantes sean atendidas por un operador. Un tiempo de espera prolongado impacta negativamente en la experiencia del cliente y es un indicador clave de ineficiencia.

   Tiempo de espera promedio = Promedio de (total_call_duration − call_duration)

3. **Volumen de llamadas salientes.**

   Cantidad total de llamadas salientes realizadas por cada operador. Este KPI sirve solo para operadores que sí deben hacer llamadas salientes y detectar a los que tienen bajo desempeño en esta tarea.

   Volumen de llamadas salientes = ∑calls_count (direction = out)
